```python
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np

# Load a blues file
y, sr = librosa.load('data/genres_original/blues/blues.00000.wav', duration=30)

# Plot waveform
plt.figure(figsize=(14, 4))
librosa.display.waveshow(y, sr=sr)
plt.title('Blues - Waveform')
plt.tight_layout()
plt.show()

# Compute and plot spectrogram
D = librosa.stft(y)
D_db = librosa.amplitude_to_db(np.abs(D), ref=np.max)

plt.figure(figsize=(14, 5))
librosa.display.specshow(D_db, sr=sr, x_axis='time', y_axis='hz', cmap='inferno')
plt.colorbar(format='%+2.0f dB')
plt.title('Blues - Spectrogram')
plt.tight_layout()
plt.show()
```

```python
genres = ['blues', 'classical', 'country', 'disco', 
          'hiphop', 'jazz', 'metal', 'pop', 'reggae', 'rock']

for genre in genres:
    y, sr = librosa.load('data/genres_original/'+genre+'/'+genre+'.00000.wav', duration=30)

    D = librosa.stft(y)
    D_db = librosa.amplitude_to_db(np.abs(D), ref=np.max)

    plt.figure(figsize=(14, 5))
    librosa.display.specshow(D_db, sr=sr, x_axis='time', y_axis='hz', cmap='inferno')
    plt.colorbar(format='%+2.0f dB')
    plt.title(genre+' - Spectrogram')
    plt.tight_layout()
    plt.show()
```

```python
fig, axes = plt.subplots(2, 5, figsize=(20, 8))
axes = axes.flatten()

for idx, genre in enumerate(genres):
    y, sr = librosa.load(f'data/genres_original/{genre}/{genre}.00000.wav', duration=30)
    D = librosa.stft(y)
    D_db = librosa.amplitude_to_db(np.abs(D), ref=np.max)
    
    librosa.display.specshow(D_db, sr=sr, x_axis='time', 
                             y_axis='hz', cmap='inferno', ax=axes[idx])
    axes[idx].set_title(genre.capitalize())
    axes[idx].label_outer()

fig.suptitle('Spectrograms by Genre', fontsize=16, y=1.02)
plt.tight_layout()
plt.savefig('spectrograms_all_genres.png', dpi=150, bbox_inches='tight')
plt.show()
```
